```python
import numpy as np

#1D

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
#INSPECTION AND PROPERTIES

print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#ARRAY OPERATIONS

temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

```python
print("Maximum Temperature:", temperatures_w1.max())
print("Minimum Temperature:", temperatures_w1.min())
print("Mean Temperature:", temperatures_w1.mean())
```

```python
#ARRAY SLICING AND INDEXING 

first_three_days = temperatures_w1[:3]

print("First three days:")
print(first_three_days)
```

```python
weekend = temperatures_w1[-2:]

print("Weekend temperatures:")
print(weekend)
```

```python
middle_three_days = temperatures_w1[2:5]

print("Middle three days:")
print(middle_three_days)
```

```python
#2D

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.1, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures:")
print(temperatures)
```

```python
#INSPECTIND AND SLICING

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
week1 = temperatures[0]
week2 = temperatures[1]

print("Week 1:")
print(week1)

print("Week 2:")
print(week2)
```

```python
week1_weekend = temperatures[0, -2:]
week2_weekend = temperatures[1, -2:]

print("Week 1 Weekend:")
print(week1_weekend)

print("Week 2 Weekend:")
print(week2_weekend)
```

```python
#PANDAS SERIES

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
#INDEXING AND SLICING

print("Mark of 1st rank student:", marks.iloc[0])
```

```python
top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]

print("Top 3 ranks:")
print(top_three)
```

```python
print("Mark of 3rd rank student:", marks.iloc[2])
```

```python
marks_above_90 = marks[marks > 90]

print("Marks greater than 90:")
print(marks_above_90)
```

```python
#MANIPULATING SERIES

marks.loc['Rank1'] = 100

print(marks)
```

```python
marks = marks.drop('Rank5')

print("After removing Rank5:")
print(marks)
```

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
#PANDAS DATAFRAME 

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Furniture', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```
